```python
# Import required libraries
import pandas as pd
import numpy as np
```

```python
# Load dataset
df = pd.read_excel("Dataset for Data Analytics.xlsx")
```

```python
# 1. Display Basic Information
print("First 5 Rows:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Info:")
print(df.info())
```

```python
# 2. Check Missing Values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# 3. Descriptive Statistics
print("\nDescriptive Statistics:")
print(df.describe())
```

```python
# 4. Calculate Mean & Median
numeric_columns = df.select_dtypes(include=np.number)

print("\nMean Values:")
print(numeric_columns.mean())

print("\nMedian Values:")
print(numeric_columns.median())
```

```python
# 5. Identify Top Products
if 'Product' in df.columns and 'TotalPrice' in df.columns:

    top_products = df.groupby('Product')['TotalPrice'].sum().sort_values(ascending=False)

    print("\nTop Selling Products:")
    print(top_products.head())
```

```python
# 6. Order Status Analysis
if 'OrderStatus' in df.columns:

    print("\nOrder Status Count:")
    print(df['OrderStatus'].value_counts())
```

```python
# 7. Outlier Detection using IQR
if 'TotalPrice' in df.columns:

    Q1 = df['TotalPrice'].quantile(0.25)
    Q3 = df['TotalPrice'].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df['TotalPrice'] < lower_bound) |
        (df['TotalPrice'] > upper_bound)
    ]

    print("\nNumber of Outliers:", len(outliers))
```
